# Release Year

```python

last_decade = netflix[["type", "release_year"]]

last_decade = last_decade.rename(columns = {"release_year" : "Release Year"})

last_decade = last_decade[last_decade["Release Year"]>=2013]
last_decade
```

```python

last_decade_df = last_decade.groupby("Release Year")["type"].size().reset_index()

last_decade_df
```

```python

last_decade_df = pd.DataFrame(last_decade_df)

last_decade_df
```

```python

last_decade_df.rename(columns = {"type": "Total Content"}, inplace = True)
```

```python

last_decade_df
```

```python

last_decade.groupby("Release Year")["type"].value_counts()
```

```python

plt.figure(figsize = (10,6))

count_plot = sns.countplot(x = "Release Year", data = last_decade, hue="type")

count_plot.set(title = "Trend of each type of content Released over the years")
```

```python

plt.figure(figsize = (10,6))

plot_total_content= sns.lineplot(x= "Release Year", y = "Total Content", data = last_decade_df)

plot_total_content.set(xlabel = "Release Year", ylabel = "Total Content", title = "Trend of content on netflix")
plt.show()
```

Result: It is seen that most content on netflix was released in year 2018, the amount of content is decreasing since then.

```python

new_catgs = {
    'TV-PG': 'Parental Guidance',
    'TV-MA' : 'Mature Audience',
    'TV-Y7-FV': 'Teens',
    'TV-Y7': 'Teens',
    'TV-14': 'Teens',
    'R': 'Mature Audience',
    'TV-Y': 'General Audience',
    'NR': 'Mature Audience',
    'PG-13': 'Teens',
    'TV-G': 'General Audience',
    'PG': 'Teens',
    'G': 'General Audience',
    'UR': 'Mature Audience',
    'NC-17': 'Mature Audience'
}

netflix['rating']=netflix['rating'].replace(new_catgs)
netflix.head()
```

```python

plt.figure(figsize= (10,6))

sns.countplot(x="rating", data=netflix, palette="pastel")

plt.title("count of Rating by Movie and Shows")
```

Result: It is seen that Netflix has more content for Mature audience.

```python
# Year-wise trend

plt.figure(figsize=(10,6))

sns.countplot(x='year_added', data=netflix, palette='pastel')

plt.title('Number of Titles Added by Year')
plt.xticks(rotation=45)
plt.show()
```

### Top Actors and Their Frequency

```python
from collections import Counter

# Drop unknown cast
cast_list = netflix[netflix['cast'] != 'unknown']['cast'].str.split(', ')
cast_list
```

```python

flat_cast = [actor for sublist in cast_list for actor in sublist]

flat_cast
```

```python

top_actors = pd.DataFrame(Counter(flat_cast).most_common(10), columns=['Actor', 'Count'])

top_actors
```

```python

sns.barplot(y='Actor', x='Count', data=top_actors, palette='pastel')

plt.title('Top 10 Most Frequent Actors on Netflix')
plt.show()
```

### Top Countries by Content Type

```python

country_type = netflix.groupby(['country', 'type']).size()

country_type
```

```python
# unstack() makes Movie and TV Show into separate columns.

country_type = country_type.unstack()

country_type
```

```python

country_type = country_type.fillna(0)
```

```python

top_countries = netflix['country'].value_counts().head(5).index

top_countries
```

```python

country_type = country_type.loc[top_countries]
country_type
```

```python

country_type.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title('Movies and TV Shows in Top 5 Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.show()
```
